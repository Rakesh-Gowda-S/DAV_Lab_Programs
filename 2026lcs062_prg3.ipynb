```python
import pandas as pandas
import matplotlib.pyplot as plt
import seaborn as sns

df=pandas.read_csv(r"C:\Users\bank\Downloads\archive (1)\Iris.csv")

```

```python
print("First 5 rows of the database:")
print(df.head())

print("\nLast 5 rows of the database:")
print(df.tail())
```

```python
print("\nShape of the database:")
print(df.shape)

print("\nColumns of the database:")
print(df.columns)
```

```python
print(df.isnull().sum())

df=df.dropna()

print(df.isnull().sum())

print(df.describe())


sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="SepalLengthCm", bins=10)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal length (cm)")
plt.ylabel("Number of Flowers")
plt.show()
```

```python

plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Species",y="SepalLengthCm")
plt.title("Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.show()
```

```python
plt.figure(figsize={8,5})
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)
plt.title("Sepal Len vs Petal len")
plt.xlabel("Sepal Length (Cm)")
```

```python
sns.pairplot(df, hue="Species")
```

```python
numeric_columns = ['Id', 'SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
correlation = df[numeric_columns].corr()
correlation
```

```python
plt.figure(figsize={8,5})
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
```
