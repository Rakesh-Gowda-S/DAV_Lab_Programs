```python
if 'student_id' not in df.columns:
    df['student_id'] = range(1, len(df) + 1)

print("\n number of duplicated student ID's")
print(df['student_id'].duplicated().sum())
```

```python
if 'grade' not in df.columns:
    df['grade'] = df['Type']

print("\nNumber of entries in each property type:")
print(df['grade'].value_counts())
```

```python
grade_info = pd.DataFrame({
    'grade': ['A', 'B', 'C', 'D', 'F'],
    'grade_description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ]
})

print("\nGrade Information DataFrame:")
print(grade_info)
```

```python
merged = pd.merge(
    df,
    grade_info,
    on='grade',
    how='left'
)

print("\nMergedDataFrame:")
print(merged.head())
```

```python
grade_details = pd.DataFrame({
    'description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ],
    'performance_level': [
        'Very High',
        'High',
        'Medium',
        'Low',
        'Very Low'
    ]
}, index=['A', 'B', 'C', 'D', 'F'])
```

```python
grade_counts = df['grade'].value_counts().to_frame('student_count')
```

```python
joined = grade_counts.join(grade_details)

print("\nJoinedDataFrame:")
print(joined)
```
